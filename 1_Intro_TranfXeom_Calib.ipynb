### Cálculo do histograma en OpenCV

OpenCV ofrece a función cv2.calcHist() para calcular o histograma:
**cv2.calcHist(images, channels, mask, histSize, ranges[, hist[, accumule]])**

* images: imaxe de orixe de tipo uint8 ou float32. Debe indicarse entre corchetes, por exemplo, “[img]”.

* channels: tamén se indica entre corchetes. É a canle para a que se calcula o histograma. Se a fonte é en escala de grises, o seu valor é [0]. Para imaxes en cor, podes pasar [0], [1] ou [2] para seleccionar a canle azul, verde ou vermella respectivamente.

* mask: imaxe de máscara. Para atopar o histograma da imaxe completa, pásase como “None”. Pero se queres atopar o histograma dunha rexión particular da imaxe, tes que crear unha imaxe binaria para indicar a rexión de interese.

* histSize: número de compartimentos. Debe indicarse entre corchetes. Para a escala completa, pasamos [256].

* intervalos: normalmente [0,256].

Comezamos cun exemplo:


```python
import numpy as np 
import cv2 
from matplotlib import pyplot as plt 
import seaborn as sns

#using matplotlib function
img = cv2.imread(intro_images_path+'pazo.jpg',0)
plt.figure(figsize=[12,4])
plt.subplot(131)
plt.title("matplotlib")
plt.ylabel('#Pixels')
plt.hist(img.ravel(),256,[0,256])

#using opencv function
hist_item = cv2.calcHist([img],[0],None,[256],[0,256]) 
#hist_norm=hist_item/img.size
plt.subplot(132)
plt.xlim([0,256])
plt.title("opencv")
plt.plot(hist_item)


plt.subplot(133)
sns.histplot(img.reshape(-1), kde=True)
#includes a smoothed version of the histogram by using KDE from the “seaborn” package 
plt.xlim([0,256])
plt.title("seaborn")

plt.show()
```

#### Translación de imaxes

A translación é o cambio da localización dun obxecto. Se coñeces o cambio na dirección (x,y) e o consideras (tx,ty), podes crear a matriz de transformación **M** do seguinte xeito:

    M=[[1, 0, tx],
       [0, 1, ty]]
   

```python
import numpy as np
from matplotlib import pyplot as plt
import cv2
import sys

# Load an color image in grayscale
img = cv2.imread(intro_images_path +'pazo.jpg')

if img is None:
    sys.exit('Failing at loading image')

img=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
num_rows, num_cols = img.shape[:2]

plt.figure(figsize=[0.01*img.shape[0],0.01*img.shape[1]])
plt.imshow(img)
plt.show()

translation_matrix = np.float32([ [1,0,70], [0,1,110] ])
#translation with image cropping
img_tc = cv2.warpAffine(img, translation_matrix, (num_cols, num_rows), cv2.INTER_LINEAR)

plt.figure(figsize=[0.01*img_tc.shape[0],0.01*img_tc.shape[1]])
plt.imshow(img_tc)
plt.show()

#translation without image cropping
img_t = cv2.warpAffine(img, translation_matrix, (num_cols+ int(translation_matrix[0,2]), num_rows+ int(translation_matrix[1,2])), cv2.INTER_LINEAR)

plt.figure(figsize=[0.01*img_t.shape[0],0.01*img_t.shape[1]])
plt.imshow(img_t)
#plt.xticks([]), plt.yticks([]) 
plt.show()
```

#### Rotación da imaxe

A rotación dunha imaxe para un ángulo θ conséguese mediante a matriz de transformación da forma

    M=[[cosθ, senθ ],
       [−sinθ, cosθ]]

OpenCV tamén proporciona unha rotación escalada con centro de rotación axustable para que poidas rotar en calquera lugar que prefiras. A matriz de transformación modificada vén dada por

    M=[[α, β, (1−α)⋅center.x−β⋅center.y],
       [-β α, β⋅center.x+(1−α)⋅center.y]]

onde:

    α=escala⋅cosθ,
    β=escala⋅senθ

Para atopar esta matriz de transformación, escribe a transformación afín ti mesmo e chama ás funcións de OpenCV **cv2.warpAffine(image, M, output_shape)**. Alternativamente, OpenCV proporciona unha función, **cv2.getRotationMatrix2D(center, angle, scale)**.

**Alternativa 1**

```python
import numpy as np
from matplotlib import pyplot as plt
import cv2 


def get_affine_cv(t, r, s):
    sin_theta = np.sin(r)
    cos_theta = np.cos(r)
    
    a_11 = s * cos_theta
    a_21 = -s * sin_theta
    
    a_12 = s * sin_theta
    a_22 = s * cos_theta
        
    a_13 = t[0] * (1 - s * cos_theta) - s * sin_theta * t[1]
    a_23 = t[1] * (1 - s * cos_theta) + s * sin_theta * t[0]
                                                          
    return np.array([[a_11, a_12, a_13],[a_21, a_22, a_23]])

#Load an image 
img = cv2.imread(intro_images_path+'pazo.jpg')
img=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    
rows,cols = img.shape[:2]
center=((cols-1)/2, (rows-1)/2)
angle=30
scale=0.6
                    
M = get_affine_cv(center, np.deg2rad(angle), scale)

img_r = cv2.warpAffine(img, M, (cols, rows))                 
plt.imshow(img_r)
plt.xticks([]), plt.yticks([])  
plt.show()
```

**Alternativa 2**

```python
img = cv2.imread(intro_images_path+'pazo.jpg')
img=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

rows,cols = img.shape[:2]
center=((cols-1)/2, (rows-1)/2)
angle=30
scale=0.6

M=cv2.getRotationMatrix2D(center, angle, scale)

img_r = cv2.warpAffine(img, M, (cols,rows))
plt.imshow(img_r)
plt.xticks([]), plt.yticks([]) 
plt.show()
```

#### Escalado de imaxes

Este é un método de uso común en visión por computador e procesamento de imaxes. Basicamente, realizamos dúas cousas no escalado de imaxes: ou ben a ampliamos ou ben a reducimos; en OpenCV, temos a función **cv2.resize(src, dsize[, dst[, fx[, fy[, interpolation]]]])**; usando isto, podemos realizar o escalado da imaxe.

Estes son os métodos de **interpolación** proporcionados por OpenCV:

- cv2.INTER_AREA usa a relación de área de píxeles para a remostraxe. Este é o mellor método para reducir o tamaño dunha imaxe. Cando se usa para ampliar a imaxe, usa o método INTER_NEAREST.

- cv2.INTER_CUBIC usa a interpolación bicúbica para o redimensionamento. Este método realiza cálculos baseados nos píxeles veciños de 4×4 para os píxeles. Toma a media ponderada destes 16 píxeles para crear un novo píxel.

- cv2.INTER_LINEAR é similar á interpolación INTER_CUBIC. Non obstante, INTER_CUBIC só usa 2×2 píxeles veciños para obter a media ponderada dos píxeles.

- cv2.INTER_NEAREST usa o concepto de veciño máis próximo para a interpolación. Isto usa só un píxel veciño da imaxe para a interpolación.

```python
import numpy as np
from matplotlib import pyplot as plt
import cv2 
import matplotlib
#matplotlib.rcParams['figure.figsize'] = (6.0, 6.0)

#Load an color image in grayscale
img = cv2.imread(intro_images_path+'pazo.jpg')
img=cv2.cvtColor(img, cv2.COLOR_BGR2RGB) 

img_shrinked = cv2.resize(img,None,fx=0.5, fy=0.5, interpolation = cv2.INTER_AREA)
plt.figure(figsize= (0.01*img_shrinked.shape[0],0.01*img_shrinked.shape[1]));plt.imshow(img_shrinked)
plt.show()

img_enlarged = cv2.resize(img,None,fx=1.5, fy=1.5, interpolation = cv2.INTER_CUBIC)
plt.figure(figsize= (0.01*img_enlarged.shape[0],0.01*img_enlarged.shape[1]));plt.imshow(img_enlarged) 
plt.show()
```

#### Obter transformacións afíns xerais a partir de correspondencias de puntos

Nas transformacións afíns, todas as liñas paralelas da imaxe de orixe seguirán sendo paralelas na imaxe de saída. Para atopar a matriz de transformación, necesitamos tres puntos da imaxe de entrada e as súas localizacións correspondentes na imaxe de saída. #### Transformación de perspectiva

A transformación de perspectiva trata da conversión do mundo 3D en imaxes 2D. O mesmo principio sobre o que funciona a visión humana e o mesmo principio sobre o que funciona a cámara.

Dadas dúas vistas 2D diferentes da mesma escena 3D, podemos estimar a transformación entre elas, simplemente identificando puntos correspondentes na mesma superficie plana.

As transformacións de perspectiva defínense mediante unha matriz de transformación completa 3x3. As liñas rectas permanecerán rectas despois da transformación. Para atopar esta matriz de transformación, necesitas 4 puntos na primeira vista e puntos correspondentes na segunda vista. Destes 4 puntos, non pode haber 3 que sexan colineais.

A matriz de transformación pódese atopar coa función **cv2.getPerspectiveTransform**. Despois, aplica **c2v.warpPerspective** con esta matriz de transformación de 3x3 para a transformación de perspectiva dunha imaxe:

**dst=cv2.warpPerspective(src, h, dsize)**
- src: imaxe de entrada.
- dst: imaxe de saída que ten o tamaño dsize e o mesmo tipo que src.
- h: homografía (matriz de transformación 3×3).
- Dsize: tamaño da imaxe de saída.

### Transformacións Proxectivas: escena 3D -> imaxe 2D

A continuación veremos as diferentes transformacións de coordenadas para pasar de coordenadas 3D nun sistema de referencia do mundo ata chegar as coordenadas imaxe de puntos sobre obxectos da escena. Consideraremos un exemplo sintético simple, puntos nos vértices dun cubo 3D, para ilustrar o proceso.


**Empezamos co código que representa un cubo nun sistema de referencia externo (mundo, e o sistema de coordenadas unha cámara ubicada nesa escena**
* T: representa a traslación da orixe sistema de coordenadas da cámara respecto do sistema de coordenadas do muno.
* R: representaa matriz de rotación en 3D do sistema de coordenadas da cámara con respecto ao sistema de coordenadas do mundo.
* P: Representa a matriz de transformación completa para coordenadas homoxéneas.

```python
import numpy as np
import matplotlib.pyplot as plt

#-------------------------Some support functions------------------------------------
def rotation_euler (theta_x, theta_y, theta_z) :
# a function that returns a 3 by 3 rotation matrix given the Euler rotation angles in radians.
    R_x = np.array([[1, 0, 0], [0, np.cos(theta_x), -np.sin(theta_x)], [0, np.sin(theta_x), np.cos(theta_x)]])
    R_y = np.array([[np.cos(theta_y), 0, -np.sin(theta_y)], [0, 1, 0], [np.sin(theta_y), 0, np.cos(theta_y)]])
    R_z = np.array([[np.cos(theta_z), -np.sin(theta_z), 0], [np.sin(theta_z), np.cos(theta_z), 0], [0, 0, 1]])

    print('R_x=\n', R_x, '\nR_y=\n', R_y, '\nR_z=\n', R_z)
    return(R_x @ R_y @ R_z)

def plot_cube (axis, p_t) :
# Given 8 vertices (p_t is 8 by 4 by 1), plot the 3D cube edges and vertices
    axis.plot3D(p_t[:,0,0], p_t[:,1,0], p_t[:,2,0], c='b', marker='o')
    axis.plot3D([p_t[0,0,0], p_t[7,0,0]], [p_t[0,1,0], p_t[7,1,0]], [p_t[0,2,0], p_t[7,2,0]], c='b', marker='o')
    axis.plot3D([p_t[2,0,0], p_t[5,0,0]], [p_t[2,1,0], p_t[5,1,0]], [p_t[2,2,0], p_t[5,2,0]], c='b', marker='o')
    axis.plot3D([p_t[1,0,0], p_t[6,0,0]], [p_t[1,1,0], p_t[6,1,0]], [p_t[1,2,0], p_t[6,2,0]], c='b', marker='o')
    axis.plot3D([p_t[0,0,0], p_t[3,0,0]], [p_t[0,1,0], p_t[3,1,0]], [p_t[0,2,0], p_t[3,2,0]], c='b', marker='o')
    axis.plot3D([p_t[4,0,0], p_t[7,0,0]], [p_t[4,1,0], p_t[7,1,0]], [p_t[4,2,0], p_t[7,2,0]], c='b', marker='o')
    axis.set_xlabel('X')
    axis.set_ylabel('Y')
    axis.set_zlabel('Z')
    axis.axis('equal')
    for i in range(8):
        axis.text(p_t[i,0,0], p_t[i,1,0], p_t[i,2,0], '{}'.format(i), 'x',fontsize='x-large')

def plot_camera (axis, c_t) :
    axis.scatter(c_t[0], c_t[1],c_t[2], c='r', marker='o')
#---------------------------------------------------------------

# Generate 8 vertices of a 5 by 5 by 5 sized cube with one of the vertices at origin of the world coordinates (homogenous)
N = 8
p_tilde = np.array([[0, 0, 0, 1], 
                    [5, 0, 0, 1], 
                    [5, 5, 0, 1], 
                    [0, 5, 0, 1], 
                    [0, 5, 5, 1], 
                    [5, 5, 5, 1], 
                    [5, 0, 5, 1], 
                    [0, 0, 5, 1]])

p_tilde = np.expand_dims(p_tilde, 2) # the dimension for p_tilde will change from (N by 4) to (N by 4 by 1)
print('Dimensions of p_tilde:', p_tilde.shape)
#---------------------------------------------------------------

T = np.array([2,2,-10])
# plot the cube, with respect to the world coordinates
fig = plt.figure(figsize=(20,10))
ax = fig.add_subplot(1,2,1, projection='3d')
plot_camera(ax,T)
ax.quiver(0,0,0,1,0,0, color='r', length=3,arrow_length_ratio=0.3)
ax.quiver(0,0,0,0,1,0, color='g', length=3,arrow_length_ratio=0.3)
ax.quiver(0,0,0,0,0,1, color='c', length=3,arrow_length_ratio=0.3)

plot_cube (ax, p_tilde)
ax.set_title('World coordinates')
#ax.view_init(azim=0, elev=0, roll=0)
#------------------------------------------------------------

# Setup the camera rotation and translation with respect to the world coordinates
R = rotation_euler(0*20*np.pi/180, 0*30*np.pi/180, 1*20*np.pi/180)

P = np.eye(4)
P[0:3,0:3] = R
P[0:3,3] = -T
print('R=\n', R, '\nT=\n', T, '\nP=\n', P)


#---------------------------------------------------------------
# Transform the 3D coordinates from world to camera coordinates
p_c_tilde = P @ p_tilde  # @ is matrix multiply in python

# P is 4 by 4 and p_tilde is 8 by 4 by 1, so this is not a standard matrix multiply!
# It is using a feature called broadcasting. It will effectively multiple each
# of the 4 by 1 homogeneous point representations by the 4 by 4 matrix to
# result in a 8 by 4 by 1 output.

'''
NUMPY BROADCASTING -- very important feature

https://www.tutorialspoint.com/numpy/numpy_broadcasting.htm

https://numpy.org/devdocs/user/basics.broadcasting.html
'''

#---------------------------------------------------------------


# plot the points in camera coordinates
ax1 = fig.add_subplot(1,2,2, projection='3d')
plot_camera(ax1,[0,0,0])

X=[1,0,0]@R
Y=[0,1,0]@R
Z=[0,0,1]@R

ax.quiver(T[0],T[1],T[2],X[0],X[1],X[2], color='r', length=3, arrow_length_ratio=0.3)
ax.quiver(T[0],T[1],T[2],Y[0],Y[1],Y[2], color='g', length=3, arrow_length_ratio=0.3)
ax.quiver(T[0],T[1],T[2],Z[0],Z[1],Z[2], color='c', length=3, arrow_length_ratio=0.3)

ax1.quiver(0,0,0,1,0,0, color='r', length=3,arrow_length_ratio=0.3)
ax1.quiver(0,0,0,0,1,0, color='g', length=3,arrow_length_ratio=0.3)
ax1.quiver(0,0,0,0,0,1, color='c', length=3,arrow_length_ratio=0.3)


plot_cube (ax1, p_c_tilde)
ax1.set_title('Camera coordinates')
plot_camera(ax1,[0,0,0])

```

## Calibración de cámaras

Para calibrar unha cámara necesitamos fixar a correspondencia entre coordenadas de puntos da escena con coordenadas píxel na imaxe capturada.

Para facer o proceso doado,emprégase habitualmente unha superficie rectangular plana que contén a imaxe dun taboleiro de xadrez. Coñecidos os tamaños do taboleiro e dos cadrados é doado identificar as esquinas na imaxe e establecer correspondencias. A partires dun conxunto de correspondencias para un conxunto de imaxes tomadas con diferentes puntos de vista e posible obter un sistema de ecuacións a partires do cal determinas os parámetros intrínsecos da cámara e a distorsión provocada pola lente.  

O código orixinal pode obterse na seguintene ligazón dun tutorial de OpenCV:

https://github.com/opencv/opencv/blob/master/samples/python/calibrate.py


Podedes descargar un plano de calibración de referencia para imprimir dende aquí:

https://www.me.psu.edu/brennan/ME545/2012/FinalProjects/Mangus_3DCameraCalibration/Calibration_Images.html

Para calibrar a vosa propia cámara teríades que imprimir esa imaxe e tomar fotos dende varios puntos de vista. Empezaremos usando unhas imaxes xa captadas e veremos como obter os parámetros de calibración


```python
import numpy as np
import cv2
from glob import glob
import matplotlib.pyplot as plt
# to run in google colab
#import sys

square_size = 2.88

img_mask = calib_images_path+'/IMG*.jpg'

#Hai que indicar o número de esquinas interiores por fila e columna (número de cadrados menos 1)
pattern_size = (7,9)

figsize = (20, 20)
```

```python

img_names = glob(img_mask)
num_images = len(img_names)
print(num_images)

pattern_points = np.zeros((np.prod(pattern_size), 3), np.float32)
pattern_points[:, :2] = np.indices(pattern_size).T.reshape(-1, 2)
pattern_points *= square_size

obj_points = []
img_points = []
h, w = cv2.imread(img_names[0]).shape[:2]
```

## Paso 1: atopar todas as esquinas no plano de calibración

```python
#plt.figure(figsize=figsize)
del_imgs=[]

for i, fn in enumerate(img_names):
    print("processing %s... " % fn)
    imgBGR = cv2.imread(fn)

    if imgBGR is None:
        print("Failed to load", fn)
        del_imgs.append(fn)
        continue

    imgRGB = cv2.cvtColor(imgBGR, cv2.COLOR_BGR2RGB)
    img = cv2.cvtColor(imgRGB, cv2.COLOR_RGB2GRAY)

    assert w == img.shape[1] and h == img.shape[0], f"size: {img.shape[1]} x {img.shape[0]}"
    found, corners = cv2.findChessboardCorners(img, pattern_size)
    # # if you want to better improve the accuracy... cv2.findChessboardCorners already uses cv2.cornerSubPix
    # if found:
    #     term = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_COUNT, 30, 0.1)
    #     cv2.cornerSubPix(img, corners, (5, 5), (-1, -1), term)

    if not found:
        print("chessboard not found")
        del_imgs.append(fn)
        continue

    if i < 12:
        img_w_corners = cv2.drawChessboardCorners(imgRGB, pattern_size, corners, found)
        plt.subplot(4, 3, i + 1)
        plt.imshow(img_w_corners)

    print(f"{fn}... OK")
    img_points.append(corners.reshape(-1, 2))
    obj_points.append(pattern_points)


plt.show()
print('Removed images: {}'.format(del_imgs))
for i, im in enumerate(del_imgs):
    img_names.remove(im)
#print(img_names)
```

## Exemplo de proxección completa de 3D a 2D dun cubo

```python
objectPoints = (
    3
    * square_size
    * np.array([[0, 0, 0], [0, 1, 0], [1, 1, 0], [1, 0, 0], [0, 0, -1], [0, 1, -1], [1, 1, -1], [1, 0, -1]])
)


def draw(img, imgpts):
    imgpts = np.int32(imgpts).reshape(-1, 2)

    # draw ground floor in green
    img = cv2.drawContours(img, [imgpts[:4]], -1, (0, 255, 0), -1)

    # draw pillars in blue color
    for i, j in zip(range(4), range(4, 8)):
        img = cv2.line(img, tuple(imgpts[i]), tuple(imgpts[j]), (255), 3)

    # draw top layer in red color
    img = cv2.drawContours(img, [imgpts[4:]], -1, (0, 0, 255), -1)

    return img


plt.figure(figsize=figsize)
for i, fn in enumerate(img_names):

    imgBGR = cv2.imread(fn)
    imgRGB = cv2.cvtColor(imgBGR, cv2.COLOR_BGR2RGB)

    dst = cv2.undistort(imgRGB, camera_matrix, dist_coefs)

    imgpts = cv2.projectPoints(objectPoints, _rvecs[i], _tvecs[i], camera_matrix, dist_coefs)[0]
    drawn_image = draw(dst, imgpts)

    if i < 12:
        plt.subplot(4, 3, i + 1)
        plt.imshow(drawn_image)

plt.show()
```

## 📷 Matriz de calibración de la cámara

La **matriz de calibración** (o matriz intrínseca, K) describe los parámetros internos de la cámara
y cómo transforma puntos 3D en coordenadas de la imagen (2D).  

Su forma general es:

\[
\begin{bmatrix}
f_x & 0   & c_x \\
0   & f_y & c_y \\
0   & 0   & 1
\end{bmatrix}
\]

donde:

- \( f_x, f_y \): Distancias focales en píxeles (eje X e Y)  
- \( c_x, c_y \): Coordenadas del punto principal (centro óptico de la imagen)
